```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
```

```python
(x_train, y_train) , (x_test, y_test) = mnist.load_data()
```

```python
print(x_train.shape)
print(y_train.shape)
print(x_test.shape)
print(y_test.shape)
```

```python
print(x_train.min())
print(x_train.max())
```

```python
unique, counts= np.unique(y_train, return_counts=True)

for digit, count in zip(unique, counts):
    print(f"Digit {digit} has {count} examples")
```

```python
plt.bar(unique,counts)
plt.xlabel('Digit')
plt.ylabel("Number of images")
plt.title("distribution of digits in the training set" )
plt.show()
```

```python
plt.figure(figsize=(10,4))
for digit in range(10):
    index=np.where(y_train==digit) [0][0]
    plt.subplot(2,5,digit+1)
    plt.imshow(x_train[index], cmap="gray")
    plt.title(f"digit: {digit}")
    plt.axis("off")
plt.tight_layout()
plt.show()
```

## Sample Images

This visualization shows one example image for each digit from 0 to 9.  
It helps confirm that the images and labels are correctly matched.

## Pixel Values

Each MNIST image is stored as a 28x28 matrix of numbers.  
Each number represents the brightness of one pixel.

- 0 means black
- 255 means white
- Values between 0 and 255 are shades of gray

This shows that images are numerical data, which is what machine learning models use.
