# Phase 3: Data Preprocessing

In this phase, I will prepare the MNIST images before training a machine learning model.

The main preprocessing steps are:

- Load the MNIST dataset
- Normalize pixel values from 0–255 to 0–1
- Reshape the image data if needed
- Prepare the labels for model training
- Confirm that the preprocessing worked correctly

```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow.keras.datasets import mnist
```

```python
(x_train, y_train) , (x_test, y_test) = mnist.load_data()
```

```python
print("Before preprocessing:")
print("Minimum pixel value:", x_train.min())
print("Maximum pixel value:", x_train.max())
```

```python
x_train_norm = x_train / 255.0
x_test_norm = x_test / 255.0
```

```python
print("After normalization:")
print("Minimum pixel value:", x_train_norm.min())
print("Maximum pixel value:", x_train_norm.max())
```

## Normalization

The original MNIST pixel values range from 0 to 255.

To prepare the images for neural network training, I normalized the pixel values by dividing them by 255.0.

After normalization:

- 0 remains 0.0
- 255 becomes 1.0
- values between 0 and 255 become decimal values between 0 and 1

This helps the model train more efficiently because the input values are smaller and more stable.

```python
x_train_cnn = x_train_norm.reshape(-1,28,28,1)
x_test_cnn = x_test_norm.reshape(-1,28,28,1)
```

```python
print("Original training shape:", x_train_norm.shape)
print("CNN training shape:", x_train_cnn.shape)

print("Original test shape:", x_test_norm.shape)
print("CNN test shape:", x_test_cnn.shape)
```

## Reshaping for CNN Input

The original MNIST images have the shape `(28, 28)`, which means each image is a 2D grayscale image.

For a Convolutional Neural Network, TensorFlow expects image data to include a channel dimension.

Before reshaping:

- Training data shape: `(60000, 28, 28)`
- Test data shape: `(10000, 28, 28)`

After reshaping:

- Training data shape: `(60000, 28, 28, 1)`
- Test data shape: `(10000, 28, 28, 1)`

The final `1` represents the grayscale color channel.

```python
print("first 10 training labels:", y_train[:10])
print("label shape:", y_train.shape)
```

## Label Preparation

The MNIST labels are already stored as integers from 0 to 9.

For this project, I will initially keep the labels in integer format because TensorFlow/Keras can train a classification model using integer labels with `sparse_categorical_crossentropy`.

This means:

- The image of digit 0 has label `0`
- The image of digit 1 has label `1`
- The image of digit 9 has label `9`

No one-hot encoding is needed at this stage.

```python
print("final training data shape:", x_train_cnn.shape)
print("final test data shape:", x_test_cnn.shape)
print("training label shape:" ,y_train.shape)
print("test label shape:" ,y_test.shape)
print("training data type:", x_train_cnn.dtype)
print("Minimum pixel value:", x_train_cnn.min())
print("Maximum pixel value:", x_train_cnn.max())
```
